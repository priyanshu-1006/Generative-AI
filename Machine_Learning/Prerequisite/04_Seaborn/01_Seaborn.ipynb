# **Seaborn**
----
### **What is Seaborn ?**
#### *Seaborn is a Data visualisation library that is built on topof matplotlib and is intergrated with Pandas.*
----

### **How Seaborn is better than Matplotlib ?**

#####  **1. Syntax** --> *Seaborn is simple --> Less lines of code is required whereas in Matplotlib a lot of lines of code are required for bringing customisation*
##### **2. By Default** --> *Seaborn Plots looks Classic and that also without even bringing any additional customization but in Matplotlib by default its plots looks similar to those of pandas plot and only after adding additional customisation in Matplotlib we can bring those visuals to life.*
##### **3. Plot Types** --> Matplotlib --> *General Plots ( Line plot, scatter plot, bar chart, histogram, pie)* --> Seaborn --> *We'll have all of the visuals for statistical analysis (KDE, box, Regression plan etc.)*
##### **4. Integration** --> *Matplotlib Integrates with all Data Types while Seaborn integrates with Pandas Dataframe.*
##### **5. Seaborn** --> *More information we can extract from the plots or visuals as compared to Matplotlib*
##### **6.** *A lot of New Plots are supported in Seaborn that are not supported in Matplotlib*

-----

## **Two Types of Graphs**

**Figure Level Graph** : *When we create a Figure Graph then the container in which that graph is created that CONTAINER --> Figure & That Graph inside the container ---> Axis*

CONTAINER --> FIGURE & GRAPH --> AXIS

*The Figure and Axis can both be refered as an Object Only just like Xlabel, ylabel etc, are also object.*
