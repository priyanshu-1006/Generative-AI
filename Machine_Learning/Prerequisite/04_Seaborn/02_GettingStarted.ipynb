## Relational Plot

Relational plot is being used to figure out the relationship between two or more numerical variables, and it helps us out to figure out how bringing changes in one numerical column affects the other numerical column.

**Relationship → Bivariate Analysis**  
**Numerical – Numerical**  
**One Numerical Column → Change → How it Affects → Other Numerical Column**

### Line Plot

A line plot is used for time series analysis: with respect to time, how one quantity is going to vary.

**Time Y axis → Value that will vary**

For example, with respect to time, how my weight is varying, how sales value is varying, or how the stock price for a company is varying.

### Scatterplot

A scatterplot shows how two numerical variables vary or are correlated. That means if one numerical column varies, it shows how it will affect the other numerical column.

**Correlation between 2 numerical columns**  
**Study Hours ↔ Exam Score**

### Axis-Level Functions

**Line plot:** `sns.lineplot()`

- `data`: Dataframe value from where we will be selecting the columns for analysis.
- `x`: `'Column_name'` that you want to keep on the X-axis. This column name will directly behave as the X-label for us.
- `y`: `'Column_name'` that you want to keep on the Y-axis. This column name will directly behave as the Y-label for us.
- `hue`: It can be group or seperate the information in the plot with respect to a new column
- `style`: Style parameter can also add additional insights in the graph by changing the plot style (lineplot --> Line style will change)

**Scatter plot:** `sns.scatterplot()`

### Figure-Level Function

**`relplot()`** is capable of plotting both scatterplot and lineplot based on the `kind` parameter:

- `kind='line'` → lineplot
- `kind='scatter'` → scatterplot

```python
## axis level function --> lineplot()

sns.lineplot(data=df, x='Age', y='CGPA',hue='Gender',style='Dietary Habits')
plt.show()
```

```python
# Figure level function --> sns.relplot(kind='line')

sns.relplot(data=df,x='Age',y='CGPA',hue='Gender', kind='line',style='Dietary Habits')
```
